```python
# Parent class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        


# Child class: Student
class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id

    def enroll_course(self, course):
        print(f"{self.name} has enrolled in {course}.")

    # Overriding the parent method
    def display_info(self):
        super().display_info()
        print(f"Student ID: {self.student_id}")


# Child class: Lecturer
class Lecturer(Person):
    def __init__(self, name, age, employee_id):
        super().__init__(name, age)
        self.employee_id = employee_id

    def teach_course(self, course):
        print(f"{self.name} is teaching {course}.")

    def display_info(self):
        super().display_info()
        print(f"Lecturer ID: {self.employee_id}")
        

# Creating a Student object
student = Student("Essel", 20, "FOE.41.006.080.25")

print("<<<Student Information>>>")
student.display_info()
student.enroll_course("Object-Oriented Programming")


# Creating a Lecturer object
lecturer = Lecturer("Dr. Matthew Cobbinah", 36, "id001")

print("<<<Lecturer Information>>>")
lecturer.teach_course("Object-Oriented Programming")
lecturer.display_info()
```
